```python
import jax.numpy as jnp
def is_ascending(x):
    n = x.shape[0]
    weights = jnp.arange(n) - (n - 1) / 2  # centered ramp: negative at start, positive at end
    return jnp.dot(weights, x)
```

```python
is_ascending(jnp.array([1,2,1]))
```
